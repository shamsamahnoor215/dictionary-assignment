Question 1: Creating & Accessing a Dictionary  

Create a dictionary named car with the following key-value pairs:  

brand: "Toyota"  

model: "Corolla"  

year: 2020

a) Access and print the brand of the car using square bracket  
notation [].  
b) Access and print the year using the .get() method.


```python
dic = {
      "brand": "Toyota",
      "model": "Corolla",
      "year": 2020
}
print("brand", dic["brand"])
print(dic.get("year"))
```

Question 5: Dictionary Methods (.keys(), .values(), .items())  
Given the dictionary:  
inventory = {"apples": 50, "bananas": 30, "oranges": 20}


a) Print all keys using .keys().  
b) Print all values using .values().  
c) Print all key-value pairs using .items().


```python
inventory = {"apples": 50, "bananas": 30, "oranges": 20}
inventory.keys()
print(list(inventory.keys()))
inventory.values()
print(list(inventory.values()))
inventory.items()
print(list(inventory.items()))
```

Question 6: Iterating through Dictionary Items  
Given the dictionary of subject marks:  
marks = {"Physics": 82, "Chemistry": 75, "Mathematics": 95, "English": 88}


Write a for loop using .items() to print each subject and its mark in the following format:  
Subject: Physics | Score: 82  
Subject: Chemistry | Score: 75



```python
marks = {"Physics": 82, "Chemistry": 75, "Mathematics": 95, "English": 88}
for subject ,score in marks.items():
  print(f"subject:{subject}| score :{score}")
```

Question 8: Conditional Logic inside a Dictionary Loop
Given a dictionary of student exam scores:  
results = {
    "Ahmad": 45,
    "Sara": 78,
    "Bilal": 92,
    "Zainab": 32,
    "Hamza": 68,
}  


Write a for loop that iterates over results.items() and:  
Prints "<Name> passed with <score>" if the score is 50 or higher.  
Prints "<Name> needs improvement (<score>)" if the score is below 50.


```python
results = {
    "Ahmad": 45,
    "Sara": 78,
    "Bilal": 92,
    "Zainab": 32,
    "Hamza": 68,
}
for i , score in results.items():
   if score>= 50:
     print("passed")
   else:
    print("needs improvement()")
```
